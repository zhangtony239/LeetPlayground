```python
class Solution:
    def exclusiveTime(self, n: int, logs: list[str]) -> list[int]:
        res_dict = {}  # 使用字典记录每个 fid 的时间
        stack = []     # 必须用栈来记录函数的调用顺序
        prev_time = 0  
        
        for log in logs:
            fid_str, status, ts_str = log.split(':')
            fid, ts = int(fid_str), int(ts_str)
            
            if status == 'start':
                if stack:
                    # 暂停当前栈顶函数，结算时间
                    curr_running = stack[-1]
                    res_dict[curr_running] = res_dict.get(curr_running, 0) + (ts - prev_time)
                stack.append(fid)
                prev_time = ts
            else:
                # 结束当前栈顶函数，弹出并结算时间
                curr_running = stack.pop()
                res_dict[curr_running] = res_dict.get(curr_running, 0) + (ts - prev_time + 1)
                prev_time = ts + 1
                
        # 因为力扣要求返回 list，我们需要按照 0 到 n-1 的顺序拼成列表
        # 这样能避免某些 fid 一直没被调用（字典里没有）而导致漏掉，没调用的默认为 0
        return [res_dict.get(i, 0) for i in range(n)]
```

```python
if __name__ == "__main__":
    solution = Solution()
    print(solution.exclusiveTime(2, ["0:start:0","1:start:2","1:end:5","0:end:6"])) # [3, 4]
```
